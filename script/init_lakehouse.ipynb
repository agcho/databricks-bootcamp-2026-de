# Select Catalog

```python
%sql
CREATE CATALOG IF NOT EXISTS bike_data;
```

# Create Lakehouse Schemas

```python
%sql
USE CATALOG bike_data;

CREATE SCHEMA IF NOT EXISTS bike_data.bronze
COMMENT 'Bronze layer: raw ingested data';

CREATE SCHEMA IF NOT EXISTS bike_data.silver
COMMENT 'Silver layer: cleaned and transformed data';

CREATE SCHEMA IF NOT EXISTS bike_data.gold
COMMENT 'Gold layer: business-ready data';
```

# Create Volume


```python
%sql
CREATE VOLUME IF NOT EXISTS bike_data.bronze.raw_sources
COMMENT 'Volume for raw source files (CSV)';
     
```
